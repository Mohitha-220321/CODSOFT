```python
import random
import string

def generate_password(length, use_upper, use_digits, use_symbols):
    characters = string.ascii_lowercase
    password = []

    # making sure atleast one character from each selected type is included
    password.append(random.choice(string.ascii_lowercase))

    if use_upper:
        characters += string.ascii_uppercase
        password.append(random.choice(string.ascii_uppercase))

    if use_digits:
        characters += string.digits
        password.append(random.choice(string.digits))

    if use_symbols:
        characters += string.punctuation
        password.append(random.choice(string.punctuation))

    # fill remaining length
    while len(password) < length:
        password.append(random.choice(characters))

    # shuffle so it doesnt look predictable
    random.shuffle(password)
    return ''.join(password)


def main():

    print("       PASSWORD GENERATOR")

    while True:
        print("\n-- Password Settings --")

        while True:
            try:
                length = int(input("Enter password length (min 6): "))
                if length >= 6:
                    break
                else:
                    print("Length must be at least 6!")
            except ValueError:
                print("Please enter a valid number!")

        use_upper   = input("Include uppercase? (y/n): ").lower() == 'y'
        use_digits  = input("Include numbers?   (y/n): ").lower() == 'y'
        use_symbols = input("Include symbols?   (y/n): ").lower() == 'y'

        password = generate_password(length, use_upper, use_digits, use_symbols)

        print("\n" + "-" * 40)
        print(f"  Generated Password : {password}")
        print(f"  Password Length    : {len(password)}")
        print("-" * 40)

        again = input("\nGenerate another password? (y/n): ").lower()
        if again != 'y':
            print("\nThank you for using Password Generator. Goodbye!")
            print("=" * 40)
            break

main()
```
